# Практическая работа №2. Разведочный анализ данных (EDA)
**Цель:** Провести базовый EDA, исследовать структуру данных, пропуски и распределения.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")
print("Библиотеки успешно загружены!")
```

## 1. Загрузка данных и первичный осмотр

```python
# Создаем тестовый датасет
data = {
    'Age': [25, 30, 35, 40, np.nan, 22, 28, 55, 18, 45],
    'Salary': [50000, 60000, 75000, 90000, 62000, 48000, 58000, 120000, 35000, 95000],
    'Department': ['IT', 'HR', 'IT', 'Finance', 'IT', 'HR', 'Finance', 'IT', 'HR', 'Finance']
}
df = pd.DataFrame(data)
df.info()
```

## 2. Анализ распределений и выбросов

```python
plt.figure(figsize=(6, 3))
sns.histplot(df['Salary'], kde=True)
plt.title('Распределение зарплат')
plt.show()
```
